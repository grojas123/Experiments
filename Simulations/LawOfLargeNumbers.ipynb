```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
n_trials = 10000
true_mean = 0.5  # Expected value
sample_means = []

# Simulate random variables (e.g., Bernoulli trials)
for n in range(1, n_trials + 1):
    sample = np.random.binomial(1, true_mean, n)
    sample_means.append(np.mean(sample))

# Plot the convergence of the sample mean
plt.figure(figsize=(10, 6))
plt.plot(range(1, n_trials + 1), sample_means, label="Sample Mean")
plt.axhline(y=true_mean, color='r', linestyle='--', label="True Mean")
plt.title("Law of Large Numbers Simulation")
plt.xlabel("Number of Trials")
plt.ylabel("Sample Mean")
plt.legend()
plt.show()
```
