```python
import math

import numpy as np
import matplotlib.pyplot as plt

# Create data points
t = np.linspace(20, 2*np.pi, 400)  # Create 400 points between 0 and 2π

# Define your functions
a = np.log(t)
b = math.exp(t)

# Plot both functions
plt.plot(t, a, 'r')  # Plot sin(t) in red
plt.plot(t, b, 'b')  # Plot cos(t) in blue
plt.show()
```
